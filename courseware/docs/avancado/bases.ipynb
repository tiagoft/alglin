# Bases

Um dos resultados mais interessantes da Álgebra Linear (pelo menos na minha opinião) é que:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = x_1 \begin{bmatrix} 1 \\ 0 \end{bmatrix} + x_2 \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

Podemos demonstrar essa propriedade:

$$
\begin{array}{rl}
 & x_1 \begin{bmatrix} 1 \\ 0 \end{bmatrix} + x_2 \begin{bmatrix} 0 \\ 1 \end{bmatrix} \\
= & \begin{bmatrix} x_1 \\ 0 \end{bmatrix} + \begin{bmatrix} 0 \\ x_2 \end{bmatrix} \\
= & \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
\end{array}
$$

E isso significa que chegar ao vetor $\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$ equivale a "andar" na direção de $\begin{bmatrix} 1 \\ 0 \end{bmatrix}$ por uma distância $x_1$, e depois $\begin{bmatrix} 0 \\ 1 \end{bmatrix}$ por uma disância $x_2$.

Só que poderíamos escolher "andar" em outras direções - e com passos que têm comprimentos diferentes "a priori". Por exemplo, poderíamos escolher:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = s_1 \begin{bmatrix} 2 \\ 3 \end{bmatrix} + s_2 \begin{bmatrix} 4 \\ 5 \end{bmatrix}
$$

E isso significa que temos duas maneiras diferentes de encontrar o vetor $\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$.

Ok, até aqui são coisas que deveríamos ter feito anteriormente no curso. Mas, agora vamos *interpretar* essas equações de uma maneira diferente.

## Vetores-base

Quando dizemos que:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = x_1 \begin{bmatrix} 1 \\ 0 \end{bmatrix} + x_2 \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

estamos dizendo que $\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$ está definido pelos vetores-base $\begin{bmatrix} 1 \\ 0 \end{bmatrix}$ e $\begin{bmatrix} 0 \\ 1 \end{bmatrix}$.

## Um vetor. Muitas bases.

Quando dizemos que:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = s_1 \begin{bmatrix} 2 \\ 3 \end{bmatrix} + s_2 \begin{bmatrix} 4 \\ 5 \end{bmatrix}
$$

estamos dizendo que o vetor $\begin{bmatrix} s_1 \\ s_2 \end{bmatrix}$ está definido pelos vetores-base $\begin{bmatrix} 2 \\ 3 \end{bmatrix}$ e $\begin{bmatrix} 4 \\ 5 \end{bmatrix}$.

## Matrizes de base

Partindo de:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = s_1 \begin{bmatrix} 2 \\ 3 \end{bmatrix} + s_2 \begin{bmatrix} 4 \\ 5 \end{bmatrix}
$$

Ao realizar a operação do lado direito da equação, chegamos a:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = \begin{bmatrix} 2s_1 + 4 s_2 \\ 3 s_1 + 5 s_2 \end{bmatrix}
$$

Veja que agora temos um sistema:

$$
\begin{cases}
\begin{array}{rl}
x_1 &= 2 s_1 + 4 s_2 \\
x_2 &= 3 s_1 + 5 s_2
\end{array}
\end{cases}
$$

e podemos escrevê-lo na forma matricial:

$$
\begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = \begin{bmatrix} 2  & 4 \\ 3 & 5 \end{bmatrix} \begin{bmatrix} s_1 \\ s_2 \end{bmatrix}
$$

Veja como os vetores-base de $\begin{bmatrix} s_1 \\ s_2 \end{bmatrix}$ se transformam nas *colunas* da matriz de coeficientes! Por isso, essa matriz é chamada de *matriz de bases* ou *matriz de base* $B$, e temos a forma compacta:

$$
x = Bs
$$



## Mudança de base

Agora podemos fazer imediatamente o seguinte exercício:

!!! info "Como escrever o vetor $\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$ na base $\begin{bmatrix} 2  & 4 \\ 3 & 5 \end{bmatrix}$?"

    Para este exercício, vamos usar imediatamente a equação:
    
    $$
    \begin{bmatrix} x_1 \\ x_2 \end{bmatrix} = \begin{bmatrix} 2  & 4 \\ 3 & 5 \end{bmatrix} \begin{bmatrix} s_1 \\ s_2 \end{bmatrix}
    $$

    *Escrever o vetor em outra base* significa encontrar esses coeficientes que multiplicam os vetores da base nova de forma a conseguirmos reconstruir nosso vetor, isto é:

    $$
    \begin{bmatrix} s_1 \\ s_2 \end{bmatrix} = 
    \begin{bmatrix} 2  & 4 \\ 3 & 5 \end{bmatrix}^{-1}  \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
    $$

??? info "Reescreva o vetor $\begin{bmatrix} 1 \\ 1 \end{bmatrix}$ na base $\begin{bmatrix} \cos(\pi/4)  & -\sin(\pi/4) \\ \sin(\pi/4) & \cos(\pi/4) \end{bmatrix}$."
    $$
    B = \begin{bmatrix} \cos(\pi/4)  & -\sin(\pi/4) \\ \sin(\pi/4) & \cos(\pi/4) \end{bmatrix} \Rightarrow B^{-1} = \begin{bmatrix} \cos(\pi/4)  & \sin(\pi/4) \\ -\sin(\pi/4) & \cos(\pi/4) \end{bmatrix}
    $$

    $$
    \begin{bmatrix}s_1\\s_2\end{bmatrix} = \begin{bmatrix} \cos(\pi/4)  & \sin(\pi/4) \\ -\sin(\pi/4) & \cos(\pi/4) \end{bmatrix} \begin{bmatrix} 1 \\ 1 \end{bmatrix}
    = \begin{bmatrix} \cos(\pi/4)  + \sin(\pi/4) \\ -\sin(\pi/4) - \cos(\pi/4) \end{bmatrix} = \begin{bmatrix} \sqrt{2} \\ 0 \end{bmatrix}
    $$

Vamos refletir sobre esse exercício.

O que fizemos foi encontrar uma maneira de expressar $x$ como a soma ponderada (ou: a combinação linear) de dois vetores-base $b_1$ e $b_2$. Em outras palavras:

$$
x = s_1 b_1 + s_2 b_2
$$

Geometricamente, o exercício se refere a esta situação:

```python
import matplotlib.pyplot as plt 
import numpy as np 
arrow_args = {
    'length_includes_head': True,
    'width': 0.01,
    'head_width': 0.1,
    'head_length': 0.1,
    'color': 'black',
}
B = np.array( [[np.cos(np.pi/4), -np.sin(np.pi/4)],
               [np.sin(np.pi/4), np.cos(np.pi/4)]] )
x = np.array([ [1], [1] ])

plt.figure(figsize=(4,4))
plt.xlim([-1.5, 1.5])
plt.ylim([-1.5, 1.5])
plt.grid()
plt.arrow(0, 0, x[0,0], x[1,0], **(arrow_args | {'color':'black'}), label='$x$' )
plt.arrow(0, 0, B[0,0], B[1,0], **(arrow_args | {'color':'red'}), label='$b_1$')
plt.arrow(0, 0, B[0,1], B[1,1], **(arrow_args | {'color':'blue'}), label='$b_2$')
plt.legend()
plt.show()
```

Então, o que significa a resposta $\begin{bmatrix} \sqrt{2} \\ 0 \end{bmatrix}$? Bem, se traçarmos um plano cartesiano em que o nosso eixo horizontal representa "apontar na direção de $b_1$" e o eixo vertical representa "apontar na direção de $b_2$", então podemos representar $x$ como:

```python
import matplotlib.pyplot as plt 
import numpy as np 
arrow_args = {
    'length_includes_head': True,
    'width': 0.01,
    'head_width': 0.1,
    'head_length': 0.1,
    'color': 'black',
}
B = np.array( [[np.cos(np.pi/4), -np.sin(np.pi/4)],
               [np.sin(np.pi/4), np.cos(np.pi/4)]] )
x = np.array([ [1], [1] ])

plt.figure(figsize=(4,4))
plt.xlim([-.5, 1.5])
plt.ylim([-.5, 1.5])
plt.grid()
plt.arrow(0, 0, np.sqrt(2), 0, **(arrow_args | {'color':'black'}), label='$x$' )
plt.arrow(0, 0, 1, 0, **(arrow_args | {'color':'red'}), label='$b_1$')
plt.arrow(0, 0, 0, 1, **(arrow_args | {'color':'blue'}), label='$b_2$')
plt.legend()
plt.show()
```
